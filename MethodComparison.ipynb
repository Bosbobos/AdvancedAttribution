# Method Comparison

Ноутбук-обёртка над `modules/MethodComparison.py`.
Можно запускать отдельные сравнения или одним вызовом собрать текущие 4 таблицы в одну директорию и один markdown-файл.
Ниже есть и тройные сравнения `IG | Cheap-IG | NAA`.


```python
from IPython.display import Markdown, display

from modules.MethodComparison import (
    compare_classifiers as _compare_classifiers_impl,
    compare_classifiers_with_cheap_ig as _compare_classifiers_with_cheap_ig_impl,
    compare_detectors as _compare_detectors_impl,
    compare_detectors_with_cheap_ig as _compare_detectors_with_cheap_ig_impl,
    export_current_notebook_tables as _export_current_notebook_tables_impl,
)


def _display_comparison_result(result):
    preview_markdown = result.get("preview_markdown")
    if preview_markdown:
        display(Markdown(preview_markdown))
    return result


def compare_classifiers(**kwargs):
    return _display_comparison_result(_compare_classifiers_impl(**kwargs))


def compare_detectors(**kwargs):
    return _display_comparison_result(_compare_detectors_impl(**kwargs))


def compare_classifiers_with_cheap_ig(**kwargs):
    return _display_comparison_result(_compare_classifiers_with_cheap_ig_impl(**kwargs))


def compare_detectors_with_cheap_ig(**kwargs):
    return _display_comparison_result(_compare_detectors_with_cheap_ig_impl(**kwargs))


def export_current_notebook_tables(**kwargs):
    return _display_comparison_result(_export_current_notebook_tables_impl(**kwargs))
```
