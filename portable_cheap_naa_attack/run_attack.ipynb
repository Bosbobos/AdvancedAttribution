# Portable Cheap-IG NAA Attack

Notebook-first entrypoint for the self-contained classifier attack package.


```python
from pathlib import Path
import sys

import matplotlib.pyplot as plt

ROOT = Path.cwd().resolve()
sys.path.insert(0, str(ROOT.parent))

from portable_cheap_naa_attack import AttackConfig, run_attack

ROOT
```

```python
image_paths = [
    str((ROOT.parent / "data" / "jess.jpg").resolve()),
]

config = AttackConfig(
    variant="base",
    batch_size=len(image_paths),
    live_plots=True,
    verbose=True,
)

config
```

```python
result = run_attack(image_paths=image_paths, config=config)
result
```

```python
print(
    f"{result.method_name}: total={result.total_runtime_seconds:.2f}s, "
    f"attribution={result.attribution_runtime_seconds:.2f}s"
)

if result.clean_importance_maps and result.final_importance_maps:
    if len(result.sample_ids) == 1:
        sample_index = 0
        sample_label = result.sample_ids[0]
    else:
        sample_index = max(range(len(result.final_target_logit_drops)), key=result.final_target_logit_drops.__getitem__)
        sample_label = result.sample_ids[sample_index]

    fig, axes = plt.subplots(1, 2, figsize=(8, 4))
    axes[0].imshow(result.clean_importance_maps[sample_index], cmap="RdBu_r", vmin=-1.0, vmax=1.0)
    axes[0].set_title(f"{result.method_name}: clean target importance ({sample_label})")
    axes[0].axis("off")
    axes[1].imshow(result.final_importance_maps[sample_index], cmap="RdBu_r", vmin=-1.0, vmax=1.0)
    axes[1].set_title(f"{result.method_name}: final target importance ({sample_label})")
    axes[1].axis("off")
    fig.suptitle(
        f"{result.method_name} | total={result.total_runtime_seconds:.2f}s | "
        f"attribution={result.attribution_runtime_seconds:.2f}s"
    )
    fig.tight_layout()
    fig
```
